```python
import csv
import random
import pandas as pd

print('Librerías cargadas correctamente')
```

```python
##Define los datos
##Lista de nombres, apellidos, ciudades y categorías académicas 
```

```python
random.seed(42)  # Fija la semilla para resultados reproducibles

nombres_m = [
    'Carlos','Luis','Andrés','Diego','Miguel','Juan','Sebastián',
    'Felipe','David','Jorge','Mateo','Nicolás','Alejandro','Ricardo',
    'Fernando','Daniel','Pablo','Cristian','Esteban','Gabriel'
]

nombres_f = [
    'María','Ana','Sofía','Valentina','Camila','Isabella','Daniela',
    'Gabriela','Fernanda','Lucía','Natalia','Carolina','Paola','Andrea',
    'Melissa','Valeria','Stephanie','Mariana','Rebeca','Karina'
]

apellidos = [
    'García','Martínez','López','Rodríguez','González','Hernández',
    'Pérez','Torres','Ramírez','Flores','Díaz','Morales','Jiménez',
    'Reyes','Álvarez','Cruz','Vargas','Castillo','Romero','Mendoza',
    'Ortiz','Herrera','Medina','Aguilar','Vega','Ríos','Soto',
    'Ramos','Núñez','Mora'
]

ciudades = [
    'Quito','Guayaquil','Cuenca','Ambato','Manta',
    'Loja','Riobamba','Santo Domingo','Ibarra','Esmeraldas'
]

jornadas = ['Matutina', 'Vespertina']
niveles  = ['Primero', 'Segundo', 'Tercero']

print(f'Nombres masculinos : {len(nombres_m)}')
print(f'Nombres femeninos  : {len(nombres_f)}')
print(f'Apellidos          : {len(apellidos)}')
print(f'Ciudades           : {len(ciudades)}')
```

```python
## Paso 3 — Función para generar notas
##Cada estudiante tiene una **base de rendimiento** (distribución normal).  
##Las notas por materia varían ligeramente alrededor de esa base.

##Escala ecuatoriana: **1–20**, mínimo para aprobar: **14**
```

```python
def generar_nota(base, offset=0.0):
    """
    Genera una nota realista basada en el rendimiento del estudiante.
    base   : rendimiento promedio del estudiante (14 = promedio nacional)
    offset : ajuste por materia (positivo = más fácil, negativo = más difícil)
    """
    nota = round(base + random.gauss(offset, 1.2), 1)
    return max(1.0, min(20.0, nota))   # Clamp entre 1 y 20

# Prueba rápida
base_ejemplo = 14.5
print('Notas de ejemplo para un estudiante con base 14.5:')
print(f'  Matemática      : {generar_nota(base_ejemplo, 0.0)}')
print(f'  Lenguaje        : {generar_nota(base_ejemplo, 0.3)}')
print(f'  Educación Física: {generar_nota(base_ejemplo, 1.0)}')
```

```python
## Paso 4 — Generar los 200 estudiantes
```

```python
filas = []

for i in range(1, 2001):

    # ── Datos personales ─────────────────────────────────────
    genero = random.choice(['M', 'F'])
    nombre = random.choice(nombres_m if genero == 'M' else nombres_f)
    apellido = f"{random.choice(apellidos)} {random.choice(apellidos)}"
    edad     = random.randint(15, 18)
    ciudad   = random.choice(ciudades)
    jornada  = random.choice(jornadas)
    nivel    = random.choice(niveles)

    # ── Rendimiento base del estudiante ──────────────────────
    # Media 14, desv. std. 2.5  →  distribución realista
    base = random.gauss(14, 2.5)
    base = max(7.0, min(20.0, base))

    # ── Notas por materia ─────────────────────────────────────
    matematica       = generar_nota(base,  0.0)
    lenguaje         = generar_nota(base,  0.3)
    ciencias         = generar_nota(base,  0.2)
    historia         = generar_nota(base,  0.5)
    ingles           = generar_nota(base, -0.2)
    educacion_fisica = generar_nota(base,  1.0)

    promedio = round(
        (matematica + lenguaje + ciencias + historia + ingles + educacion_fisica) / 6, 2
    )

    # ── Asistencia (correlacionada con rendimiento) ───────────
    # Mejor promedio → tiende a faltar menos
    asistencia_pct   = round(min(100.0, max(50.0, random.gauss(85 + (promedio - 14) * 1.5, 6))), 1)
    clases_totales   = 180
    clases_asistidas = round(clases_totales * asistencia_pct / 100)
    clases_faltadas  = clases_totales - clases_asistidas

    aprobado = 'Sí' if promedio >= 14 else 'No'

    filas.append({
        'id_estudiante'   : i,
        'nombre'          : nombre,
        'apellido'        : apellido,
        'genero'          : genero,
        'edad'            : edad,
        'ciudad'          : ciudad,
        'jornada'         : jornada,
        'nivel'           : nivel,
        'matematica'      : matematica,
        'lenguaje'        : lenguaje,
        'ciencias'        : ciencias,
        'historia'        : historia,
        'ingles'          : ingles,
        'educacion_fisica': educacion_fisica,
        'promedio_general': promedio,
        'asistencia_pct'  : asistencia_pct,
        'clases_asistidas': clases_asistidas,
        'clases_faltadas' : clases_faltadas,
        'aprobado'        : aprobado
    })

print(f'Estudiantes generados: {len(filas)}')
print(f'Ejemplo fila 1: {filas[0]}')
```

```python
## Paso 5 — Verificar con Pandas antes de guardar
```

```python
df = pd.DataFrame(filas)

print('=== Primeras 5 filas ===')
display(df.head())

print('\n=== Tipos de datos ===')
print(df.dtypes)

print('\n=== Valores nulos ===')
print(df.isnull().sum())
```

```python
print('=== Estadísticas descriptivas (notas) ===')
cols_notas = ['matematica','lenguaje','ciencias','historia','ingles','educacion_fisica','promedio_general']
display(df[cols_notas].describe().round(2))

print('\n=== Distribución por género ===')
print(df['genero'].value_counts())

print('\n=== Aprobados vs Reprobados ===')
print(df['aprobado'].value_counts())

print('\n=== Estudiantes por nivel ===')
print(df['nivel'].value_counts())
```

```python
## Paso 6 — Guardar el archivo CSV
```

```python
campos = [
    'id_estudiante','nombre','apellido','genero','edad','ciudad',
    'jornada','nivel','matematica','lenguaje','ciencias','historia',
    'ingles','educacion_fisica','promedio_general','asistencia_pct',
    'clases_asistidas','clases_faltadas','aprobado'
]

nombre_archivo = 'data/estudiantes.csv'

with open(nombre_archivo, 'w', newline='', encoding='utf-8') as f:
    writer = csv.DictWriter(f, fieldnames=campos)
    writer.writeheader()
    writer.writerows(filas)

print(f'Archivo guardado: {nombre_archivo}')
print(f'Filas escritas  : {len(filas)}')
print(f'Columnas        : {len(campos)}')
```
